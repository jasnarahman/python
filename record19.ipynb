```python
filea=input("Enter the source file")
fileb=input("Enter the destination file")
f1=open(filea,"r")
f2=open(fileb,"w")
i=0 
for line in f1.readlines():
    i=i+1
    if i%2==0:
        continue
    f2.write(line)
f1.close()
f2.close()
print("Odd lines copied successfully from",filea,"to",fileb)
```

```python
import csv
fname=input("Enter csv file name")
with open (fname,"r")as f1:
    reader=csv.DictReader(f1)
    print("\nDepartment Details")
    print("......................")
    for row in reader:
        print("DeptId:",row['DeptId'],"\tDeptName",row['DeptName'])
        
```

```python
import csv
data=[{"Name":"Anu","Age":22,"Marks":85},{"Name":"Arun","Age":21,"Marks":78},{"Name":"jasna","Age":30,"Marks":90}]
with open("studentsdict.csv","w",newline='')as file:
    fieldnames=["Name","Age","Marks"]
    writer=csv.DictWriter(file,fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data)
print("dictionary data written sccessfully")
with open("studentsdict.csv","r")as file:
    reader=csv.DictReader(file)
    for row in reader:
        print(row)
```
